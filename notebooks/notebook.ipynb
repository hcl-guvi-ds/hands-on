```python
# decorator
def my_decorator(func):
    def wrapper():
        print("Starting")
        func()
        print("ending")
    return wrapper
```

```python
@my_decorator
def print_name():
    print("hello")
```

```python
print_name()
```

```python
# decorator with arguments

def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("This is argument", args)
        print("This is kwargs", kwargs)
        result = func(*args, **kwargs)
        print("result printed")

        return result 
    return wrapper
        
```

```python
@my_decorator
def my_add(a, b):
    print(a+b)
```

```python
my_add(10, 20)
```

```python
def wrapper(*args, **kwargs):
    print("This is argument", args)
    print("This is kwargs", kwargs)
    print(kwargs['name'], kwargs['age'])

wrapper(2, 3, name='Amit', age=24)
```
