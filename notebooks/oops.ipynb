```python
class BankAccount:
    def __init__(self, account_no, account_holder_name, balance, account_type):
        self.__account_no = account_no
        self.__account_holder_name = account_holder_name
        self.__account_type = account_type
        self.__balance = balance
    def deposit(self, amount):
        self.__balance += amount
        print("Deposited:", amount)
    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient Balance")
    def display(self):
        print("Account Number :", self.__account_no)
        print("Account Holder Name :", self.__account_holder_name)
        print("Balance :", self.__balance)
        print("Account Type : ",self.__account_type)
acc1 = BankAccount(1001, "Anitha", 5000, "SB")
acc1.display()
acc1.deposit(2000)
acc1.withdraw(1500)
acc1.display()
```

```python
# Multiwheeler  RoadTransport ---> FourWheeler ---> Car

class RoadTransport:

    def __init__(self, wheel, engine):
        self.wheel = wheel
        self.engine = engine 
    
    def start_engine(self):
        print(f"{self.engine} has started") 

class FourWheeler(RoadTransport):

    def __init__(self, wheel, engine, steering, wipers):
        self.steering = steering 
        self.wipers = wipers 
        super().__init__(wheel, engine)     # calling the contructor of the parent class

    def start_engine(self):
        print(f"Four wheeler {self.engine} has started")

class Car(FourWheeler):

    def __init__(self, name, wheel, engine, steering, wipers):
        self.name = name
        super().__init__(wheel, engine, steering, wipers)
        print(f"Name of the Car is {self.name}")
    
    def fuel_type(self):
        print(f"Fuel type of {self.name} is petrol")
```

```python
c = Car('Porche',4, 'Ford', True, True)
c.start_engine()
```

```python
f = FourWheeler(4, 'Ford', True, True)
f.start_engine()
```

```python
def my_decorator(func):

    def wrapper():
        print(" Starting the log, function execution started")
        func()
        print("Function execution is over")
    
    return wrapper
```

```python
greet = my_decorator(greet)
greet()
```

```python
@my_decorator   # it's same as saying welcome = my_decorator(welcome)
def welcome():
    print("Welcome to the decorators")
```

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Function is Starting")
        result = func(*args, **kwargs)
        print("Function execution is over")
        return result     
    return wrapper
```

```python
@decorator
def add(a, b, **kwargs):
    return [a+b, kwargs['name']]
```

```python
print(add(10, 12, name="Amit", age=24))
```
