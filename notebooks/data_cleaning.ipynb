```python
import numpy as np
def convert_size(size):
    if pd.isna(size):
        return np.nan
    if size == 'Varies with device':
        return np.nan
    if size.endswith('M'):
        return float(size[:-1])
    if size.endswith('k'):
        return float(size[:-1])/1024
    
    return np.nan

df['Size'] = df['Size'].apply(convert_size)
```

```python
def clean_text(text):
    text = text.lower()
    text = text.strip()
    return text
```
